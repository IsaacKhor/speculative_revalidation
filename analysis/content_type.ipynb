```python
import duckdb, numpy
con = duckdb.connect(database=':memory:')
def q(query: str):
    return con.execute(query).df()
```

```python
q('''
describe '../traces/cf/**/*.parquet'
''')
```

```python
df = q('''
select count(*) as n, mime
from '../traces/cf/**/*.parquet'
group by mime
having n > 100000
order by n desc
  ''')
df.head(50)
```
